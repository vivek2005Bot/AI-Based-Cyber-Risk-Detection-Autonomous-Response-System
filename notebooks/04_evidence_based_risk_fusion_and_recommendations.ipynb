# 04 - Evidence-Based Risk Fusion & Response Recommendation
## AI-Based Cyber Risk Detection System

This notebook demonstrates:
1. **Mathematical Evidence Fusion**: Integrating supervised ML probabilities, unsupervised anomaly scores, and contextual security indicators (CVSS v3, asset criticality, threat intel, decay).
2. **Interpretable Risk Scoring**: Deriving continuous $[0, 1]$ risk scores and mapped severity tiers (Minimal, Low, Medium, High, Critical).
3. **Response Recommenders**:
   - Content-Based (TF-IDF cosine similarity)
   - Collaborative Filtering (Matrix Factorization)
   - Hybrid Recommender with incident response plans.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from src.risk_fusion.fusion import RiskFusionEngine
from src.risk_fusion.contextual import ContextualIndicatorEngine
from src.recommendation.content_based import ContentBasedRecommender
from src.recommendation.collaborative import CollaborativeRecommender
from src.recommendation.hybrid import HybridRecommender

print("Risk fusion & recommendation engines ready!")
```

### 1. Evidence-Based Risk Fusion Simulation

```python
fusion_engine = RiskFusionEngine()

# Simulate 5 distinct threat scenarios
supervised_probs = np.array([0.05, 0.40, 0.65, 0.88, 0.95])
anomaly_scores   = np.array([0.10, 0.30, 0.70, 0.80, 0.92])
contextual_scores = np.array([0.02, 0.25, 0.50, 0.75, 0.90])

fused_scores = fusion_engine.fuse(supervised_probs, anomaly_scores, contextual_scores)
risk_levels = fusion_engine.risk_level(fused_scores)

for i in range(len(fused_scores)):
    explanation = fusion_engine.explain_risk(fused_scores[i], supervised_probs[i], anomaly_scores[i], contextual_scores[i])
    print(f"Scenario {i+1}: Risk Score = {fused_scores[i]:.3f} [{risk_levels[i].upper()}] | Dominant: {explanation['dominant_factor']}")
```

### 2. Mitigation Recommendation Approaches

```python
hybrid_rec = HybridRecommender()

event = {
    "event_type": "network",
    "risk_level": "critical",
    "attack_type": "ddos_attack",
    "cvss_score": 9.8,
}

recs = hybrid_rec.recommend(event, top_k=4)
print(f"Recommended Tactical Responses for {event['attack_type'].upper()}:")
for i, r in enumerate(recs, 1):
    print(f" {i}. Action: {r['action'].upper():<20} | Score: {r['score']:.3f} | {r['description']}")
```
