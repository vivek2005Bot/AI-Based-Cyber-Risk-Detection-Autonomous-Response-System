### 1. Simulating Environment & Training Q-Learning Agent

```python
agent = QLearningAgent()
print(f"State Dimension: {agent.state_dim} | Actions Available: {agent.n_actions}")
print(f"Action Space: {agent.env.get_action_names()}")

# Train for 200 episodes
metrics = agent.train(n_episodes=200)
eval_metrics = agent.evaluate(n_episodes=50)

print(f"\nEvaluation Win Rate: {eval_metrics['win_rate']*100:.1f}%")
print(f"Mean Evaluation Reward: {eval_metrics['mean_reward']:.2f}")

plt.figure(figsize=(8, 4))
plt.plot(metrics['episode_rewards'], color='forestgreen', alpha=0.7)
plt.title('Q-Learning Episode Rewards Convergence')
plt.xlabel('Episode')
plt.ylabel('Cumulative Reward')
plt.grid(True, alpha=0.3)
plt.show()
```

### 2. Multi-Agent Coordinated Tactical Analysis

```python
mas = MultiAgentSystem()
sim_event_state = np.array([0.90, 0.85, 0.70, 0.80, 0.75, 0.95])

analysis = mas.analyze_event(sim_event_state)
print("Specialist Signals:")
for sig in analysis["signals"]:
    print(f" - {sig['agent']:<18} ({sig['focus']}): Signal={sig['signal']:.3f}, Confidence={sig['confidence']:.2f}")

print(f"\nCentral Decision Agent Selected Strategy: {analysis['recommended_action'].upper()}")
```
